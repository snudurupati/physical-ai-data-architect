## Local PySpark 4 and Jupyter Notebooks 

```python
import os, glob
from pyspark.sql import SparkSession

os.environ["JAVA_HOME"] = "/opt/homebrew/opt/openjdk@17"

spark = (SparkSession
            .builder
                .master("local[*]")
                .appName("paida")
                .config("spark.driver.memory", "16g")
                .getOrCreate())
```

```python
# 1. Expand the tilde (~/) to an absolute path string
search_pattern = os.path.expanduser("~/.cache/huggingface/lerobot/lerobot/pusht/data/*/*.parquet")

# 2. Find all matching file paths on your local machine
local_files = glob.glob(search_pattern)

df = spark.read.parquet(*local_files)

df.select("episode_index").distinct().count()
```
